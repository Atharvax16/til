```python
import numpy as np
```

```python
RNG = np.random.default_rng(0)
#fixed seed --> reproducible (scientific practice)
```

```python
def make_synthesis(n = 500,d = 3,noise = 0.5):
    X_raw = RNG.normal(size = (n,d))
    X = np.hstack([np.ones((n,1)),X_raw])
    w_true = RNG.normal(size = d + 1)
    y = X @ w_true + noise * RNG.normal(size = n)
    return X,y,w_true
```

```python
def train_test_split(X,y,test_frac = 0.2):
    n = X.shape[0]
    idx = RNG.permutation(n)
    cut = int(n * (1 - test_frac))
    tr,te = idx[:cut],idx [cut:]
    return X[tr],y[tr],X[te],y[te]
```

```python
def normal_equation(X, y):
    return np.linalg.solve(X.T @ X, X.T @ y)
```
